```python
import pandas as pd
import numpy as np
import joblib
from src.utils.config import RAW_DATA_PATH, PROCESSED_DATA_PATH, MODEL_PATH

df = pd.read_csv(PROCESSED_DATA_PATH)
df.head()
```

```python
print(f"Total records: {len(df)}")
print(f"Overall no-show rate: {df['no_show'].mean():.2%}")
df.groupby('appointment_weekday')['no_show'].mean()
```

```python
model = joblib.load(MODEL_PATH)
sample_input = df.iloc[:5].drop(columns=['appointment_id', 'booking_date', 'appointment_date', 'no_show'])
probs = model.predict_proba(sample_input)[:, 1]
for i, p in enumerate(probs):
    print(f"Sample {i+1} predicted no-show probability: {p:.2%}")
```
